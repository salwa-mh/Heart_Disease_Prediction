```python
import pandas as pd
import joblib
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix, precision_score, recall_score, f1_score


train = pd.read_csv("../data/processed/data_with_scalling/cleaned_train_Scaling.csv")
test = pd.read_csv("../data/processed/data_with_scalling/cleaned_test_Scaling.csv")

X_train = train.drop("Heart Disease", axis=1)
y_train = train["Heart Disease"]

X_test = test.drop("Heart Disease", axis=1)
y_test = test["Heart Disease"]

kernels = ['linear', 'rbf']
C_values = [0.1, 1, 10]


for k in kernels:
    for c in C_values:

        
        print(f"SVM Model | Kernel = {k} | C = {c}")
        print("-" * 50)
        model = SVC(kernel=k, C=c, probability=True)

        model.fit(X_train, y_train)

        
        y_pred = model.predict(X_test)

        
        acc = accuracy_score(y_test, y_pred)
        pre = precision_score(y_test, y_pred, zero_division=0)
        rec = recall_score(y_test, y_pred, zero_division=0)
        f1 = f1_score(y_test, y_pred, zero_division=0)
        cm = confusion_matrix(y_test, y_pred)
        print("Accuracy :", acc)
        print("Precision:", pre)
        print("Recall   :", rec)
        print("F1 Score :", f1)

        print("\nConfusion Matrix:")
        print(cm)

        print("\n")
        
        joblib.dump(model, f"../models/svm_{k}_C{c}.pkl")
        joblib.dump(cm, f"../results/cm_{k}_C{c}.pkl")
        joblib.dump(y_pred, f"../results/y_pred_{k}_C{c}.pkl")
        joblib.dump(y_test, f"../results/y_test_{k}_C{c}.pkl")
        
    
        if k == "rbf" and c == 1:
            joblib.dump(y_test, "../results/y_test.pkl")
            joblib.dump(y_pred, "../results/y_pred.pkl")
            joblib.dump(model, "../models/svm_model.pkl")
            joblib.dump(cm, "../results/cm.pkl")





joblib.dump(X_train.columns, "../models/columns.pkl")
joblib.dump(X_train.sample(50, random_state=42), "../models/background.pkl")


```
