```python
import pandas as pd

train = pd.read_csv("../data/processed/data_with_scalling/cleaned_train_Scaling.csv")
test = pd.read_csv("../data/processed/data_with_scalling/cleaned_test_Scaling.csv")

X_train = train.drop("Heart Disease", axis=1)
y_train = train["Heart Disease"]

X_test = test.drop("Heart Disease", axis=1)
y_test = test["Heart Disease"]

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

models = [
    DecisionTreeClassifier(max_depth=3),
    DecisionTreeClassifier(max_depth=5),
    DecisionTreeClassifier(max_depth=10),
    DecisionTreeClassifier(max_depth=None)
]

results = []

for i, model in enumerate(models):
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred)
    rec = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)

    results.append([i+1, acc, prec, rec, f1])

    print("Model", i+1)
    print(confusion_matrix(y_test, y_pred))
    print("\n")



df_results = pd.DataFrame(results, columns=[
    "Model", "Accuracy", "Precision", "Recall", "F1-score"
])

print(df_results)
print("\n")

# from sklearn.tree import DecisionTreeClassifier
# from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# models = [
#     DecisionTreeClassifier(max_depth=3),
#     DecisionTreeClassifier(max_depth=5),
#     DecisionTreeClassifier(max_depth=10),
#     DecisionTreeClassifier(max_depth=None)
# ]

# for i, model in enumerate(models):
#     model.fit(X_train, y_train)
#     y_pred = model.predict(X_test)

#     acc = accuracy_score(y_test, y_pred)
#     prec = precision_score(y_test, y_pred)
#     rec = recall_score(y_test, y_pred)
#     f1 = f1_score(y_test, y_pred)

#     print("Model", i+1)
#     print("Accuracy:", round(acc, 3))
#     print("Precision:", round(prec, 3))
#     print("Recall:", round(rec, 3))
#     print("F1-score:", round(f1, 3))
#     print("\n")
```
