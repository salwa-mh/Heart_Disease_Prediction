```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

 #read  test/train
train = pd.read_csv("../data/processed/data_with_scalling/cleaned_train_Scaling.csv")
test = pd.read_csv("../data/processed/data_with_scalling/cleaned_test_Scaling.csv")


#split  train data into features&target
x_train = train.drop("Heart Disease", axis=1)
y_train=train["Heart Disease"]


#split  test data into features&target
x_test = test.drop("Heart Disease", axis=1)
y_test = test["Heart Disease"]


n_estimators= [50, 100, 200]
max_depth = [5, 10, None]
#  list results
results = []
#model number
id = 1
for n in n_estimators:
    for d in max_depth:
        # create Random Forest model
        model = RandomForestClassifier(
            n_estimators=n,
            max_depth=d,
            min_samples_split=5,
            min_samples_leaf=2,
            random_state=42)
        #model train
        model.fit(x_train, y_train)
         #model prediction of test data
        y_pred = model.predict(x_test)

        #calaulate evaluation metrices
        acc = round(accuracy_score(y_test, y_pred), 3)
        prec = round(precision_score(y_test, y_pred), 3)
        rec = round(recall_score(y_test, y_pred), 3)
        f1 = round(f1_score(y_test, y_pred), 3)
        #confusion matrix values
        tn, fp, fn, tp = confusion_matrix(y_test, y_pred).ravel()
        #saving results of each model
        results.append([id, n, d, acc, prec, rec, f1, tn, fp, fn, tp])
         #increase model number(next model)
        id += 1
#all models results>>table
results_table = pd.DataFrame(results, columns=[ "id","trees_num","max_depth", "accuracy","precision","recall","f1","TN","FP","FN","TP"])

print("All experiment results:")
#print all result,best f1,confusion matrix
print(results_table[["id","trees_num","max_depth", "accuracy","precision","recall","f1"]])
best_f1 = results_table["f1"].max()
print("\nBest F1 Score =", best_f1)
print("\nConfusion Matrix Values:\n")
print(results_table[[ "id", "trees_num","max_depth","TN", "FP","FN","TP"]])
```
