```python
import matplotlib.pyplot as plt


smok = [
    dt['smoking_status_formerly smoked'].sum(),
    dt['smoking_status_never smoked'].sum(),
    dt['smoking_status_smokes'].sum(),
    dt['smoking_status_Unknown'].sum()
]

labels = ['formerly smoked', 'never smoked', 'smokes', 'unknown']

plt.figure(figsize=(6,6))
plt.pie(smok, labels=labels, startangle=90, autopct='%1.1f%%')

plt.title("Smoking Status Distribution")
plt.show()
```

```python
import matplotlib.pyplot as plt


smok = [
    df['smoking_status_formerly smoked'].sum(),
    df['smoking_status_never smoked'].sum(),
    df['smoking_status_smokes'].sum(),
    df['smoking_status_Unknown'].sum()
]

labels = ['formerly smoked', 'never smoked', 'smokes', 'unknown']

plt.figure(figsize=(6,6))
plt.pie(smok, labels=labels, startangle=90, autopct='%1.1f%%')

plt.title("Smoking Status Distribution")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data_plot = dt.groupby('Heart Disease')[[
    'smoking_status_formerly smoked',
    'smoking_status_never smoked',
    'smoking_status_smokes',
    'smoking_status_Unknown'
]].sum().T


data_plot.plot(kind='bar', figsize=(8,5))

plt.title('Heart Disease Prevalence by Smoking Status')
plt.xlabel('Smoking Status')
plt.ylabel('Count')
plt.show()
```

```python
work_types = [
    'work_type_Private',
    'work_type_Self-employed',
    'work_type_Govt_job',
    'work_type_Never_worked'
]

data = dt.groupby('Heart Disease')[work_types].sum().T

data.plot(kind='bar', figsize=(8,5))

plt.title("Impact of Work Type on Heart Disease")
plt.xticks(rotation=45)
plt.ylabel("Count")
plt.show()
```
