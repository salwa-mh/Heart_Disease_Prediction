```python
import pandas as pd
from sklearn.naive_bayes import GaussianNB
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.model_selection import cross_validate, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix




train_data = pd.read_csv("../data/processed/data_with_scalling/cleaned_train_Scaling.csv")
test_data  = pd.read_csv("../data/processed/data_with_scalling/cleaned_test_Scaling.csv")

X_train = train_data.drop("Heart Disease", axis=1)
y_train = train_data["Heart Disease"]
X_test = test_data.drop("Heart Disease", axis=1)
y_test = test_data["Heart Disease"]

X_train = X_train.loc[:, X_train.nunique() > 1]
X_test = X_test.loc[:, X_train.columns]

# Find best k (Feature Selection step)
best_k = 0
best_score = 0

for k in range(5, 19):

    pipe = Pipeline([
        ('select', SelectKBest(score_func=f_classif, k=k)),
        ('model', GaussianNB())
    ])

    score = cross_val_score(pipe, X_train, y_train, cv=5, scoring='f1').mean()

    print("k =", k, "F1 =", score)
    print("\n")

    if score > best_score:
        best_score = score
        best_k = k

print("\nBest k =", best_k)
print("\n")



# Apply best feature selection(the best k)
selector = SelectKBest(score_func=f_classif, k=best_k)

X_train_sel = selector.fit_transform(X_train, y_train)
X_test_sel = selector.transform(X_test)



# var_smoothing with the best k value
values = [1e-9, 1e-8, 1e-7]

for v in values:

    model = GaussianNB(var_smoothing=v)

    model.fit(X_train_sel, y_train)
    y_pred = model.predict(X_test_sel)

    print("\nvar_smoothing: ", v,)
    print("Accuracy :", accuracy_score(y_test, y_pred))
    print("Precision:", precision_score(y_test, y_pred))
    print("Recall   :", recall_score(y_test, y_pred))
    print("F1-score :", f1_score(y_test, y_pred))
    print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))



# Cross Validation (Final Model) (after optimization steps)

pipe = Pipeline([
    ('select', SelectKBest(score_func=f_classif, k=15)),
    ('model', GaussianNB(var_smoothing=1e-8))
])

print("\nCross Validation:")

cv_results = cross_validate(
    pipe,
    X_train,
    y_train,
    cv=5,
    scoring=['accuracy', 'precision', 'recall', 'f1']
)

print("CV Accuracy:", cv_results['test_accuracy'].mean())
print("CV Precision:", cv_results['test_precision'].mean())
print("CV Recall:", cv_results['test_recall'].mean())
print("CV F1:", cv_results['test_f1'].mean())
```
